# Requête API
Notebook to fetch NASA datasets.
Use of python library `earthaccess` to fetch data from earthdata.nasa.gov

**Documentation:**

https://pypi.org/project/earthaccess/ <br>
https://earthaccess.readthedocs.io/en/latest/user/quick-start/

```python
import os
os.system("conda install -c conda-forge earthaccess")

import earthaccess
```

```python
os.environ["EARTHDATA_USERNAME"] = ""
os.environ["EARTHDATA_PASSWORD"] = ""

earthaccess.login(strategy="environment")

results = earthaccess.search_data(
    short_name="MODIS_AQUA_L3_SST_THERMAL_MONTHLY_9KM_NIGHTTIME_V2019.0",
    temporal=("2002-07-01", "2003-10-31"),
)

earthaccess.download(results, "./data")
```
